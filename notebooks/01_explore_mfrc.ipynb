```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("USC-MOLA-Lab/MFRC")
print(ds)
df = ds[list(ds.keys())[0]].to_pandas()
print(df.shape, df.columns.tolist())
df.head(5)
```

```python
print("unique texts:", df["text"].nunique())
print("annotators per text:\n", df.groupby("text")["annotator"].nunique().value_counts().sort_index())
print("max rows per (text, annotator):", df.groupby(["text", "annotator"]).size().max())
print("texts in >1 subreddit:", (df.groupby("text")["subreddit"].nunique() > 1).sum())

u = df.drop_duplicates("text")
print(u["subreddit"].value_counts(), "\n", u["bucket"].value_counts())
print(df["annotation"].str.split(",").explode().str.strip().value_counts())
print(df["confidence"].value_counts(dropna=False))
```
